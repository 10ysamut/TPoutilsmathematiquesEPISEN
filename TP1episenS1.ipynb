</div>

La factorielle d'un entier naturel $n$ intervient partout : en probabilités et en dénombrement, en informatique, en physique statistique, etc.

Le problème est que $n!$ devient très vite gigantesque : $10!$ vaut déjà plus de trois millions, et $100!$ possède $158$ chiffres. Il est donc naturel de chercher une **formule simple** qui donne l'ordre de grandeur de $n!$ lorsque $n$ est grand. C'est l'objet de la **formule de Stirling** :

  <h1>Problème 2 - Suites récurrentes et chaos</h1>
</div>

De nombreux phénomènes évoluent par étapes : chaque état se déduit du précédent par une règle fixe. On les modélise par des **suites récurrentes** de la forme $u_{n+1} = f(u_n)$. C'est le cas, par exemple, de l'évolution d'une population d'une année sur l'autre, de la propagation d'une épidémie ou de certains algorithmes itératifs.

Dans ce problème, nous étudions l'une des plus célèbres : la **suite logistique**, introduite en biologie pour modéliser une population animale. Pour $r \in [0,4]$, on pose $f_r(x) = r\,x(1-x)$ et
$$u_0 \in [0,1], \qquad u_{n+1} = r\,u_n(1-u_n).$$
Ici, $u_n$ représente la population l'année $n$ (normalisée entre $0$ et $1$) et $r$ le taux de reproduction. Le terme $1 - u_n$ traduit le fait que, lorsque la population devient trop importante, les ressources manquent et la croissance ralentit.

Malgré sa formule très simple, cette suite a un comportement étonnamment riche selon la valeur de $r$ :
1. nous commencerons par observer numériquement la suite pour différentes valeurs de $r$ ;
2. nous démontrerons ensuite sa convergence dans deux cas particuliers ;
3. nous visualiserons son comportement à l'aide des diagrammes en toile d'araignée ;
4. enfin, nous découvrirons le **chaos** : une suite parfaitement déterministe peut devenir imprévisible, et une infime modification de $u_0$ peut changer complètement son évolution.

Pour visualiser le comportement d'une suite récurrente $u_{n+1} = f_r(u_n)$, on utilise un **diagramme en toile d'araignée**. On trace sur un même graphique la courbe $y = f_r(x)$ et la droite $y = x$, puis on construit les termes de la suite graphiquement :
- partant du point $(u_0, u_0)$ situé sur la droite $y = x$, on trace un segment **vertical** jusqu'à la courbe : on arrive au point $(u_0, u_1)$ ;
- on trace ensuite un segment **horizontal** jusqu'à la droite $y = x$ : on arrive au point $(u_1, u_1)$ ;
- on recommence à partir de $(u_1, u_1)$, et ainsi de suite.
